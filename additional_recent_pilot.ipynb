# Additional real-review feasibility pilot: Civilization VI

**November 15, 2026 — Public Steam Community page.** This examines 10 available real negative reviews, with paraphrases only, for coding feasibility. **This is not the official API sample.** Source: https://steamcommunity.com/app/289070/negativereviews/

## Review summary records

The source page displays posting days/months without years; we infer 2026 from the page context and **cannot guarantee dates without API validation**. Page playtime may be newer than review-creation playtime. `F0/F1/F2/F3` are preliminary manual codes; `EXCLUDED` means blank text.

```python
import pandas as pd
from pathlib import Path
path = Path("latest_page_10_paraphrased.csv")
latest = pd.read_csv(path)
latest[["sample_id","posted_date_inferred","feedback_level","issue_topics","summary_paraphrased"]]
```

## Small-sample descriptive summaries (not population estimates)

We report the 2026-10 subset separately from the November reviews, because the A5 time window ends October 31.

```python
in_scope = latest[latest["in_2026_may_oct_window"].astype(str).str.lower() == "true"]
print("All visible reviews:", len(latest))
print("Apparently within May–Oct 2026:", len(in_scope))
print("All 10 feedback levels:")
print(latest.feedback_level.value_counts().to_string())
print("\nFive October feedback levels:")
print(in_scope.feedback_level.value_counts().to_string())
```

```python
import matplotlib.pyplot as plt
order = ["F0","F1","F2","F3","EXCLUDED"]
values = [int((in_scope.feedback_level == t).sum()) for t in order]
fig, ax = plt.subplots(figsize=(7,3.4))
ax.bar(order, values)
ax.set_ylabel("Count (n=5)")
ax.set_xlabel("Feedback code")
ax.set_title("October 2026 visible negative reviews: feasibility check")
ax.set_ylim(0,max(values)+1)
plt.tight_layout()
plt.show()
```

## Implications for the full study

- The classification distinction is usable for this small set: general judgment (F0), specific complaints (F1), and explicit change requests (F2) all appear. No proposed implementation details (F3) appear in these October reviews.
- Product policy/pricing and accessibility/UI problems should be explicit codebook topics; not all feedback is game mechanics.
- Reviews can be blank and must have a documented exclusion rule.
- These visible reviews cannot establish rates for English Steam-purchaser negative reviews, and the published date needs API confirmation.
