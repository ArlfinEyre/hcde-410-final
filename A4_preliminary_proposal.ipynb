## 1. Motivation & Problem

Steam reviews are often summarized as positive or negative recommendations, but the written comments can contain more detailed feedback about game design, technical issues, and desired improvements. I am interested in how players use negative reviews not only to evaluate a game, but also to point out problems and propose changes.

**Preliminary research question:** **To what extent do negative Steam reviews of *Sid Meier's Civilization VI* identify specific problems or suggest improvements, and what kinds of issues do they address?**

I chose *Civilization VI* as a provisional case because it is a complex strategy game with an established Steam review community. This study could show what kinds of information are present behind a binary rating, and where reviews may be useful—or limited—as a form of player feedback. I will study what reviewers explicitly write, **not assume their intentions or judge whether their suggestions would actually work**.


## 2. Data

**Primary source:** Steam's public **IUserReviewsService.GetAppReviews** API for *Sid Meier's Civilization VI* (Steam App ID **289070**).

- API documentation: https://partner.steamgames.com/doc/webapi/IUserReviewsService
- Game page: https://store.steampowered.com/app/289070/
- API terms: https://steamcommunity.com/dev/apiterms

**Relevant fields:** review text (`review`), recommendation (`voted_up`), review language, creation and update timestamps, playtime at review (`author.playtime_at_review`), and optional developer responses. These fields allow me to identify negative reviews, describe the sample, and analyze what feedback the written comments contain.

**Access and licensing:** The reviews are publicly accessible through Steam, but this does **not** mean the review text is open-licensed. Valve's Steam Web API Terms govern access and reuse. The MIT LICENSE in my repository applies only to my original notebook/code, **not** to Steam comments or other third-party content.

**Ethics and representativeness:** Reviewers are self-selected and do not represent all players. I will avoid publishing SteamIDs, usernames, or a bulk raw-text review archive; share reproducible procedures and aggregated results instead. Reviews may be sarcastic, edited, or context-dependent. Review language is not a reliable proxy for nationality.


## 3. Tentative Analysis Approach

I plan to combine **quantitative description** of Steam reviews with a smaller **qualitative content analysis**:

1. **Pilot:** Retrieve approximately 30–50 reviews to verify API access, field availability, text quality, and whether the proposed categories are useful.
2. **Quantitative:** Collect a bounded, documented sample of negative reviews (initial target: several hundred) and summarize review dates, lengths, and playtime-at-review distributions.
3. **Qualitative coding:** Systematically select a smaller set (tentatively 100–150) for manual coding. Initial distinctions: (a) general dissatisfaction, (b) a specific problem, and (c) an explicit request or suggestion for change. Possible issue topics include interface/usability, gameplay rules/balance, technical problems, and content/updates. A review can receive multiple codes.
4. **Integrate:** Report the frequency of coded categories **within the coded sample**, alongside selected anonymized paraphrases and discussion of ambiguous cases. Refine definitions after the pilot rather than assuming the initial codebook is final.

The main study will focus on **one game** and initially **one language**, with a Chinese-language comparison or 2–3 short interviews only if they clearly add value and are feasible under course research-ethics expectations. No interviews are necessary to complete the primary analysis.


## 4. Unknowns & Dependencies

- **API/data feasibility:** I have not yet collected the reviews. Steam may change access or impose rate limits; the pilot will verify that the current API works for this study.
- **Scope and sampling:** I need to choose a consistent date range and sampling procedure. Ranking by “most helpful” alone could bias the sample toward longer or unusually popular reviews.
- **Coding reliability:** The distinction between a complaint and a specific suggestion can be subjective. I will define categories clearly, test them on a pilot sample, and permit overlapping or unclear cases.
- **Game context:** DLC, updates, and changing player expectations may affect what reviews discuss. I will document the sampled period and avoid causal claims.
- **Fallback:** If improvement suggestions are too rare or inconsistently identifiable, I can narrow the study to the types of **specific problems** expressed in negative reviews (especially interface/usability vs. game-mechanics concerns), while keeping the same data source and case game.

**Peer feedback:** Classmates asked how I would obtain and work with the data, what analysis methods I would use, and how I would narrow the question. In response, I am focusing on one game, starting with a pilot, and proposing explicit text-coding categories rather than inferring psychological motivations from playtime.
