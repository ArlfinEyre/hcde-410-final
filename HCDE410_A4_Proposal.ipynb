# A4: Final Project Preliminary Proposal


## 1. Motivation & Problem

Steam reviews have a simple thumbs-up or thumbs-down rating, but the written comments can say much more. Some players just say they dislike a game, while others point out specific problems or suggest changes. I'm interested in whether these reviews can be useful feedback for game developers, beyond helping other people decide what to buy.

**My research question:** How often do negative Steam reviews of *Civilization VI* describe specific problems or suggest changes, and what kinds of issues do they focus on?

I'm starting with *Civilization VI* because it has complex game systems and a large Steam review community.

## 2. Data

I'll use Steam's public review API for *Civilization VI* (App ID **289070**). The data includes review text, positive/negative recommendations, playtime, review dates, and languages. This should let me look at both overall patterns and what people actually write.

- **Data source and documentation:** https://partner.steamgames.com/doc/webapi/IUserReviewsService
- **Game:** https://store.steampowered.com/app/289070/
- **Terms:** https://steamcommunity.com/dev/apiterms

The reviews are public, but they aren't automatically free to redistribute. I'll follow Steam's terms, avoid sharing reviewer identities, and publish code and summarized findings rather than a full collection of comments. Steam reviewers are also not representative of everyone who plays the game. The MIT license in my repo applies to my own work, not Steam review data.

## 3. Planned Analysis

First, I'll test a small sample of about 30–50 reviews to see whether the API works and whether the comments are useful for this question.

Then I'll collect a larger, documented sample of negative reviews. I'll use Python to summarize things like review dates and playtime. For a smaller sample of comments, I'll code whether players:

- Give a general complaint
- Describe a specific problem
- Suggest a change or improvement

I'll also note what the feedback is about, such as game mechanics, interface, technical issues, or updates. I'll compare the coded themes and read selected reviews more closely to understand the patterns. These categories can overlap.

For now, I'll focus on one game. A Chinese-language comparison or a few short interviews would only be optional extensions.
