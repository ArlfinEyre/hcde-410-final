### Manual coding and consistency

The [codebook](CODEBOOK_PILOT.md) has definitions and **illustrative examples, not real review quotations**. I will code two dimensions for each review:

**1. Feedback specificity (one highest supported level):**
- `F0` — General negative judgment without an identifiable issue.
- `F1` — Identifies a concrete issue, without explicitly asking for change.
- `F2` — Requests an improvement or a changed outcome.
- `F3` — Suggests a particular remedy or implementation.

For a simpler chart, I can combine these as **general dissatisfaction (F0)**, **specific problems (F1)**, and **improvement suggestions (F2+F3)**, but I will preserve the four detailed codes. A higher code does not automatically mean the feedback is correct, actionable, or helpful. Blank comments are counted as exclusions; genuinely ambiguous comments are flagged for review.

**2. Issue topics (multiple labels allowed):**
- **Game systems:** AI opponent behavior, gameplay balance, diplomacy, combat and units, city/district management, pacing and progression, or other mechanics.
- **Other issues:** UI/tutorials/accessibility, crashes and performance, patches/mod compatibility, privacy/terms, launcher, DLC/pricing, and other concerns.

The distinction between **AI opponents** and **balance** matters: an unintelligent AI decision and an unfair numerical difficulty bonus are not necessarily the same complaint. Labels may overlap (for example, a complaint about AI diplomacy can receive both AI and diplomacy tags). I will not code the professor's guess that AI is most criticized as an expected result.

After developing the rules on the pilot, I will code the full sample. Then I will **re-code 15–20% of the reviews after a break**, compare the first and second decisions, record disagreements, and revise unclear definitions. Because there is only one coder, I will call this a **within-coder/code–recode consistency check**, not inter-rater reliability. I will keep a short log of codebook changes.

### API collection code for the revised five-period plan

This code is included as a **reproducible plan** but has **not** completed live collection here. It uses the current `IUserReviewsService/GetAppReviews` endpoint, sends filters via `input_json`, and requests each date window separately. Start with `RUN_LIVE_PILOT = True` when running in an environment with Steam access. Raw comments and review IDs stay under `data_private/` (excluded by `.gitignore`).

The code checks page completion before allowing a stratified random sample. Anonymous requests can be rate-limited (HTTP 429); do not use repeated aggressive calls to bypass Steam limits.

```python
import json
import time
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd
import requests

APP_ID = 289070
ENDPOINT = "https://api.steampowered.com/IUserReviewsService/GetAppReviews/v1/"
RUN_LIVE_PILOT = False  # Turn on only in a network-enabled Python environment

PERIODS = {
    "base_2017": ("2017-03-01", "2017-08-31"),
    "rise_fall_2018": ("2018-03-01", "2018-08-31"),
    "gathering_2019": ("2019-03-01", "2019-08-31"),
    "frontier_2020": ("2020-06-01", "2020-11-30"),
    "recent_2026": ("2026-05-01", "2026-10-31"),
}

def utc_bounds(dates):
    start, end = dates
    begin = int(datetime.fromisoformat(start + "T00:00:00+00:00").timestamp())
    finish = int(datetime.fromisoformat(end + "T23:59:59+00:00").timestamp())
    return begin, finish

def fetch_period(period_name, max_pages=1, page_size=10, wait_seconds=1.5):
    """Fetch one defined period. No claim of complete retrieval until verified."""
    first, last = utc_bounds(PERIODS[period_name])
    config = {
        "appid": APP_ID,
        "filter": 1,                        # Recently created, not Helpful ranking
        "review_type": 2,                   # Negative
        "purchase_type": 0,                 # Steam purchasers
        "languages": ["english"],
        "date_range_start": first,
        "date_range_end": last,
        "filter_offtopic_activity": True,
        "num_per_page": min(page_size, 100),
        "cursor": "*",
    }
    seen = set()
    rows = []
    total_matching = None
    completed = False
    for page in range(max_pages):
        for attempt in range(3):
            response = requests.get(
                ENDPOINT, params={"input_json": json.dumps(config)}, timeout=30
            )
            if response.status_code == 429 and attempt < 2:
                time.sleep(5 * (2 ** attempt))
                continue
            response.raise_for_status()
            break
        payload = response.json()
        body = payload.get("response", payload)
        batch = body.get("reviews", [])
        if body.get("total_matching") is not None:
            total_matching = int(body["total_matching"])

        if not batch:
            completed = True
            break
        for item in batch:
            rid = str(item.get("recommendationid", ""))
            if not rid or rid in seen:
                continue
            seen.add(rid)
            created = item.get("timestamp_created")
            if (created is None or not (first <= int(created) <= last)
                    or item.get("voted_up") is not False
                    or item.get("language") != "english"
                    or item.get("steam_purchase") is False):
                continue
            updated = int(item.get("timestamp_updated") or created)
            play_min = (item.get("author") or {}).get("playtime_at_review")
            rows.append({
                "period": period_name,
                "review_id": rid,           # Private only
                "review": item.get("review", ""),
                "created_utc": datetime.fromtimestamp(int(created), tz=timezone.utc).isoformat(),
                "updated_utc": datetime.fromtimestamp(updated, tz=timezone.utc).isoformat(),
                "edited_since_creation": updated > int(created),
                "playtime_at_review_hours": (
                    None if play_min is None else float(play_min) / 60
                ),
            })
        if total_matching is not None and len(seen) >= total_matching:
            completed = True
            break
        next_cursor = body.get("cursor")
        if not next_cursor or next_cursor == config["cursor"]:
            # Could be an API paging issue; only trust a full count or empty final page
            completed = (total_matching is not None and len(seen) >= total_matching)
            break
        config["cursor"] = next_cursor
        time.sleep(wait_seconds)
    return pd.DataFrame(rows), {
        "period": period_name, "pages_requested_max": max_pages,
        "seen_ids": len(seen), "matching_reported": total_matching,
        "complete_frame": completed,
    }

if RUN_LIVE_PILOT:
    Path("data_private").mkdir(exist_ok=True)
    pilot_parts = []
    pilot_status = []
    try:
        for label in PERIODS:
            part, status = fetch_period(label, max_pages=1, page_size=10)
            pilot_parts.append(part)
            pilot_status.append(status)
            time.sleep(1.5)
        pilot = pd.concat(pilot_parts, ignore_index=True)
        pilot.to_csv("data_private/pilot_reviews_private.csv", index=False)
        print("Pilot retrieved:", len(pilot), "reviews across", len(PERIODS), "periods")
        print(pd.DataFrame(pilot_status).to_string(index=False))
        print("Review and code the actual text; do not commit the private CSV.")
    except requests.RequestException as exc:
        print("Live API unavailable; no representative results:", type(exc).__name__, str(exc)[:300])
else:
    print("No live Steam request was made. Set RUN_LIVE_PILOT=True to test.")
```

### Planned period-stratified main sample

Run the full collector **only after the 40–50-review pilot succeeds** and the codebook has been checked against actual text. Collect each date window to completion (not only the first 100 reviews). Use fixed seed 410 and select up to 40 eligible, nonblank reviews per period; report the number available and any exclusions. If one period's frame is incomplete, stop rather than pretending that a sample from its first pages is random.

The code uses a conservative page cap, so a period may still be truncated; an incomplete flag means the full design must be revised or collected in smaller date sub-windows. All raw text and IDs must remain **private**, not on public GitHub.

```python
RUN_FULL_COLLECTION = False  # Enable only after a successful live pilot and API check

if RUN_FULL_COLLECTION:
    Path("data_private").mkdir(exist_ok=True)
    period_frames = []
    collection_status = []
    for period_name in PERIODS:
        frame, status = fetch_period(period_name, max_pages=250, page_size=100)
        period_frames.append(frame)
        collection_status.append(status)
        time.sleep(1.5)

    status_table = pd.DataFrame(collection_status)
    print(status_table.to_string(index=False))
    if not status_table["complete_frame"].all():
        print("STOP: at least one period frame is incomplete. Do not report a random stratified sample.")
    else:
        sampled_parts = []
        for i, frame in enumerate(period_frames):
            eligible = frame.loc[frame["review"].fillna("").str.strip().ne("")].copy()
            take = min(40, len(eligible))
            selected = eligible.sample(n=take, random_state=410 + i) if take else eligible
            sampled_parts.append(selected)
            print(list(PERIODS)[i],
                  "eligible:", len(eligible), "sampled:", take)
        coded_sample = pd.concat(sampled_parts, ignore_index=True)
        coded_sample.to_csv("data_private/coding_sample_private.csv", index=False)
        status_table.to_csv("data_private/collection_status_private.csv", index=False)
        print("Prepared", len(coded_sample), "records for manual coding (PRIVATE).")
else:
    print("Full collection not run; no formal sample or estimates claimed.")
```

**Still to do:** an independent delayed code–recode of 15–20% of the coded sample; complete per-period sampling frames and randomly select about 40 from each period; then conduct the final qualitative and quantitative analysis. This first pass is not a reliability result and is not the final 200-review study.
