## 5. Overview & Research Questions

**Working title:** *Beyond Thumbs Down: What Negative Steam Reviews Tell Us About Player Feedback*

Steam's thumbs-down label tells us that a reviewer does not recommend a game, but not what they want other players or the developer to know. I want to look more closely at the **information inside negative reviews**, using *Sid Meier's Civilization VI* as a single case. Its many game systems make it a useful setting for studying feedback about rules, usability, balance, and technical issues.

**Main research question:** To what extent do negative Steam reviews of *Civilization VI* describe specific problems or suggest improvements, and what issues do they discuss?

**Supporting questions:**

1. What proportion of sampled negative reviews express only general dissatisfaction, describe an identifiable problem, or request/propose a change?
2. Which problem categories (e.g., gameplay, interface, AI/balance, technical issues, or DLC/pricing) come up in these reviews, and how are they connected to the level of detail in the feedback?

This is an **exploratory, descriptive study**, not a test of what causes negative reviews or whether the developer should implement particular suggestions. The unit of analysis is **one review**, not an individual player's mental state. I will start with English-language negative reviews of one game; Chinese-language comparisons and interviews are not required for the main project.

## 6. Background & Related Work

Research on Steam reviews suggests that player comments serve a purpose beyond a simple purchase recommendation. Tong (2021) studied *No Man's Sky* and argued that game reviews can help developers identify problems and player needs during ongoing development. This supports treating reviews as potential feedback, but it does not mean every review contains a useful suggestion.

Tong, Willcock, and Sun (2024; published in the March 2025 journal issue) compared several topic-modeling approaches on *No Man's Sky* reviews and interviewed seven developers about the usefulness of the topics. Their work shows both the value of identifying patterns in many reviews and the importance of interpreting those patterns for an actual game context. Their later study (Tong et al., 2025) also highlights that Steam reviews can mix multiple kinds of information. For my smaller study, I will use a **transparent manual coding scheme** instead of a complex topic model, so that the distinction between a complaint, a specific problem, and a suggestion stays easy to inspect.

Hsieh and Shannon (2005) distinguish conventional, directed, and summative qualitative content analysis. Their discussion supports starting with a small set of practical coding categories, revising them after reading real comments, and combining counts with interpretation of context.

**How my study adds to this work:** I am not trying to show for the first time that Steam reviews can inform developers. Instead, I will ask a narrower question: *what kinds of concrete feedback actually appear in the negative reviews of one complex strategy game, and how specific is that feedback?* This is also a way to think critically about how much information a binary rating leaves out.

### Review coding

I will manually code each sampled review on **two separate dimensions**:

**Feedback specificity (one highest applicable level per review):**

- `F0` — Only general dissatisfaction; no identifiable issue.
- `F1` — A specific issue is identified, but no clear request for change.
- `F2` — A change or improvement is explicitly requested.
- `F3` — A concrete implementation idea or proposed solution is described.

A review mentioning several points gets the **highest** level supported by its text. `F3` means specific, **not** technically feasible or objectively good. Sarcasm or unclear meaning may be coded `unclear` and discussed rather than forced into a level.

**Issue topics (multiple labels allowed):** `gameplay_rules`, `ai_balance`, `ui_onboarding`, `technical_performance`, `privacy_terms`, `launcher`, `dlc_pricing`, `other`, plus `no_identifiable_issue` where appropriate. A review can mention multiple topics. I will refine the labels after the pilot and keep a codebook with definitions and borderline examples. I will re-code 15–20% of comments after a short break to check my own consistency, and report disagreements and any rule changes; this is **not** inter-rater reliability.

### Analysis and presentation

I will use Python/Pandas to summarize the collected reviews (collection period, review lengths, playtime-at-review distributions, missing/edited fields). For the manually coded sample, I will report counts and percentages of `F0`–`F3` and topic labels. I will create (1) a bar chart of feedback specificity, (2) a bar chart of issue topics (multiple labels possible), and (3) a simple cross-tab of specificity and topic. Then I will close-read a small set of informative and ambiguous comments to explain the distinctions behind the counts. Any quotations will be brief and appropriately attributed or paraphrased, consistent with data-use terms.

The main conclusion will describe **what sampled reviewers chose to express**, not whether their feedback is correct, what their hidden motivations are, or what all *Civilization VI* players believe. ### API pilot: reproducible collection code

The code below is **included for reproducibility but has not been run against live Steam data here**. Set `RUN_LIVE_PILOT = True` when you have an internet connection. By default it will not contact Steam. It writes raw review text only under `data_private/`, which is excluded from Git by `.gitignore`.

It uses the **new** `IUserReviewsService` endpoint and the `input_json` parameter (not the deprecated `/appreviews` URL). If Steam changes the API or responds with rate limits, check the official documentation before retrying.

```python
import json
import time
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd
import requests

APP_ID = 289070  # Civilization VI
ENDPOINT = "https://api.steampowered.com/IUserReviewsService/GetAppReviews/v1/"
START_UTC = datetime(2026, 5, 1, tzinfo=timezone.utc)
END_UTC = datetime(2026, 10, 31, 23, 59, 59, tzinfo=timezone.utc)
RUN_LIVE_PILOT = False  # Change to True to test Steam API


def fetch_reviews(max_pages=1, page_size=40, wait_seconds=1.2):
    """Retrieve pages of English negative reviews; return records and paging status."""
    config = {
        "appid": APP_ID,
        "filter": 1,                   # Recent by creation date
        "review_type": 2,              # Negative
        "purchase_type": 0,            # Purchased on Steam
        "languages": ["english"],
        "date_range_start": int(START_UTC.timestamp()),
        "date_range_end": int(END_UTC.timestamp()),
        "num_per_page": min(page_size, 100),
        "filter_offtopic_activity": True,
        "cursor": "*",
    }
    all_records = []
    seen_ids = set()
    complete = False
    total_matching = None
    for _ in range(max_pages):
        response = requests.get(
            ENDPOINT,
            params={"input_json": json.dumps(config)},
            timeout=30,
        )
        response.raise_for_status()  # Includes 429 rate limits
        payload = response.json()
        body = payload.get("response", payload)
        batch = body.get("reviews", [])
        total_matching = body.get("total_matching", total_matching)
        if not batch:
            complete = True
            break
        for review in batch:
            review_id = str(review.get("recommendationid", ""))
            if not review_id or review_id in seen_ids:
                continue
            seen_ids.add(review_id)
            created = review.get("timestamp_created")
            if (created is not None
                    and int(START_UTC.timestamp()) <= created <= int(END_UTC.timestamp())
                    and review.get("voted_up") is False
                    and review.get("language") == "english"):
                all_records.append({
                    "review_id": review_id,  # Private only; not for GitHub
                    "review": review.get("review", ""),
                    "created_utc": datetime.fromtimestamp(created, tz=timezone.utc).isoformat(),
                    "updated_utc": datetime.fromtimestamp(
                        review.get("timestamp_updated", created), tz=timezone.utc
                    ).isoformat(),
                    "playtime_at_review_hours":
                        review.get("author", {}).get("playtime_at_review", 0) / 60,
                })
        if total_matching is not None and len(seen_ids) >= total_matching:
            complete = True
            break
        next_cursor = body.get("cursor")
        if not next_cursor or next_cursor == config["cursor"]:
            complete = True
            break
        config["cursor"] = next_cursor
        time.sleep(wait_seconds)
    return pd.DataFrame(all_records), complete, total_matching


if RUN_LIVE_PILOT:
    try:
        pilot_df, pilot_complete, pilot_total = fetch_reviews(max_pages=1, page_size=40)
        Path("data_private").mkdir(exist_ok=True)
        pilot_df.to_csv("data_private/pilot_reviews_private.csv", index=False)
        print("Pilot records retrieved:", len(pilot_df))
        print("Fields:", list(pilot_df.columns))
        print("Snapshot is private; do not commit the CSV.")
        print("Live retrieval succeeded. Manually inspect and code reviews before reporting theme frequencies.")
    except requests.RequestException as error:
        print("Live API test failed; no pilot findings are available:", type(error).__name__, str(error)[:200])
else:
    print("Live pilot NOT run; set RUN_LIVE_PILOT = True when online.")
```

Keep all raw text and any review IDs in the local `data_private/` directory. Do not commit it to GitHub. Coding categories should be tested against real text before final quantitative analysis.

```python
RUN_FULL_COLLECTION = False  # Leave False until pilot succeeds and scope is confirmed

if RUN_FULL_COLLECTION:
    full_df, is_complete, total_matching = fetch_reviews(max_pages=100, page_size=100)
    print("Retrieved:", len(full_df), "API total_matching:", total_matching,
          "Complete frame:", is_complete)
    if not is_complete:
        print("WARNING: frame not complete; do not make population claims or treat sample as random.")
    else:
        eligible = full_df[full_df["review"].notna()].copy()
        n = min(200, len(eligible))
        sample = eligible.sample(n=n, random_state=410) if n else eligible
        Path("data_private").mkdir(exist_ok=True)
        sample.to_csv("data_private/coding_sample_private.csv", index=False)
        print("Prepared", n, "reviews for manual coding (private local file).")
else:
    print("Full collection not run; no counts or findings are claimed.")
```
